```python
import pandas as pd
import string
```

```python
df = pd.read_csv("./train.csv")
df.head()
```

```python
num_classes = df['category'].nunique()
print(num_classes)
```

```python
df['text'].head()
```

```python
def lower(text):
    text = text.lower()
    return text


df['text'] = df['text'].apply(lower)
df.head()
```

```python
def punc(text):
    text = text.translate(str.maketrans("","", string.punctuation))
    return text

df['text'] = df['text'].apply(punc)
df.head()
```

```python
from sklearn.model_selection import train_test_split

x = df['text']
y = df['category']


xtrain, xtest, ytrain, ytest = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vect = TfidfVectorizer(ngram_range=(1, 2))

# xtrain = vect.fit_transform(xtrain)
# ytrain = vect.transform(ytrain)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(xtrain, ytrain)
```

```python

from sklearn.metrics import accuracy_score, classification_report

pred = model.predict(xtest)

print("Accuracy:", accuracy_score(ytest, pred))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vect = TfidfVectorizer(ngram_range=(1, 2))

xtraina = vect.fit_transform(xtrain)
xtesta = vect.transform(xtest)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

ytraina = encoder.fit_transform(ytrain)
ytesta = encoder.transform(ytest)

num_classes = len(encoder.classes_)
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras  import layers
from tensorflow.keras.layers import Input, Dense, Dropout
```

ANN1 -> input -> 16 -> 8 -> Output 

```python
ann1 = keras.Sequential([
    layers.Input(shape=(xtraina.shape[1],)),
    layers.Dense(16, activation="relu"),
    layers.Dense(8, activation="relu"),
    layers.Dense(num_classes, activation="softmax")
])
```

```python
ann1.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python

history1 = ann1.fit(
    xtraina.astype("float32").toarray(),
    ytraina,
    validation_data=(
        xtesta.astype("float32").toarray(),
        ytesta
    ),
    epochs=100,
    batch_size=64,
    verbose=1
)

#val_accuracy: 0.7281
```

ANN2 128, 64, 32, output

```python
ann2 = keras.Sequential([
    layers.Input(shape=(xtraina.shape[1],)),

    layers.Dense(256, activation="relu"),
    layers.Dropout(0.3),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.2),

    layers.Dense(num_classes, activation="softmax")
])

ann2.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

earlystop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)
```

```python
history2 = ann2.fit(
    xtraina.astype("float32").toarray(),
    ytraina,
    validation_data=(
        xtesta.astype("float32").toarray(),
        ytesta
    ),
    epochs=25,
    batch_size=64,
    verbose=1,
    callbacks=[earlystop]
)
```

```python
ann3 = keras.Sequential([
    layers.Input(shape=(xtraina.shape[1],)),

    layers.Dense(256, activation="relu"),
    layers.Dropout(0.4),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.3),

    layers.Dense(num_classes, activation="softmax")
])

ann3.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

earlystop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=4,
    restore_best_weights=True
)

history3 = ann3.fit(
    xtraina.astype("float32").toarray(),
    ytraina,
    validation_split=0.15,
    epochs=25,
    batch_size=64,
    verbose=1,
    callbacks=[earlystop]
)

#126/126 ━━━━━━━━━━━━━━━━━━━━ 4s 32ms/step - accuracy: 0.9929 - loss: 0.0402 - val_accuracy: 0.8866 - val_loss: 0.4526
```

```python
import os
import joblib

os.makedirs("banking77_model", exist_ok=True)

# Save the model and vectorizer from the SAME run
ann3.save("banking77_model/ann.keras")
joblib.dump(vect, "banking77_model/tfidf.pkl")
joblib.dump(encoder, "banking77_model/label_encoder.pkl")

print("Model expects:", ann3.input_shape[1])
print("Training matrix:", xtraina.shape[1])
print("Current vectorizer:", len(vect.vocabulary_))
```
